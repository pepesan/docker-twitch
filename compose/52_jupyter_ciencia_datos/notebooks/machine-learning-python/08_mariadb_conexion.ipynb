# Conexión a MariaDB del laboratorio

El servicio Compose `mariadb` crea la base `ml_lab` y el usuario local `ml_lab`. Desde Jupyter se conecta por el nombre de servicio y lee credenciales de variables de entorno. La consulta siguiente no modifica datos. No publiques credenciales ni expongas MariaDB fuera de la red privada de Compose.


```python
import os
import mysql.connector

connection = mysql.connector.connect(
    host=os.environ.get("MARIADB_HOST", "mariadb"),
    port=int(os.environ.get("MARIADB_PORT", "3306")),
    database=os.environ.get("MARIADB_DATABASE", "ml_lab"),
    user=os.environ.get("MARIADB_USER", "ml_lab"),
    password=os.environ.get("MARIADB_PASSWORD", "ml_lab_local_only"),
)
try:
    with connection.cursor() as cursor:
        cursor.execute("SELECT DATABASE(), VERSION()")
        print(cursor.fetchone())
finally:
    connection.close()
```
