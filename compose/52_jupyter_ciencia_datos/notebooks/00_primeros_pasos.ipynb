# Primeros pasos en análisis de datos

Ejemplo reproducible con datos sintéticos. Usa NumPy y pandas para preparar datos, Seaborn y Matplotlib para visualizarlos, y scikit-learn para entrenar un modelo sencillo.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn

print(f"NumPy {np.__version__} | pandas {pd.__version__} | SciPy {scipy.__version__} | scikit-learn {sklearn.__version__}")
```

```python
rng = np.random.default_rng(42)
n = 120
horas_estudio = rng.uniform(0, 10, n)
asistencia = rng.uniform(0.5, 1.0, n)
nota = 2 + 0.55 * horas_estudio + 3 * asistencia + rng.normal(0, 0.8, n)

datos = pd.DataFrame({
    'horas_estudio': horas_estudio,
    'asistencia': asistencia,
    'nota': np.clip(nota, 0, 10).round(2),
})
datos.head()
```

```python
datos.describe().round(2)
```

```python
sns.regplot(data=datos, x='horas_estudio', y='nota', scatter_kws={'alpha': 0.5})
plt.title('Relación entre horas de estudio y nota')
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

X = datos[['horas_estudio', 'asistencia']]
y = datos['nota']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
modelo = LinearRegression().fit(X_train, y_train)
predicciones = modelo.predict(X_test)
print(f'MAE: {mean_absolute_error(y_test, predicciones):.2f}')
print(f'R²: {r2_score(y_test, predicciones):.2f}')
```
